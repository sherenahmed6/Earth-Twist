```python
# Count geographic points
print("Geographic Points:",
      df.groupby(["LAT", "LON"]).ngroups)
# Records per year
print("\nRecords per year:")
print(df.groupby("YEAR").size())
# Records per geographic point
print("\nRecords per point:")
print(df.groupby(["LAT", "LON"]).size().value_counts())
```
