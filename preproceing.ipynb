```python
precipitation_long_w = precipitation_raw.melt(
    id_vars=["YEAR", "LAT", "LON"],
    value_vars=[
        "JAN", "FEB", "MAR", "APR",
        "MAY", "JUN", "JUL", "AUG",
        "SEP", "OCT", "NOV", "DEC"
    ],
    var_name="MONTH",
    value_name="PRECIPITATION"
)
```

```python
precipitation_long_w
```

```python
precipitation_long_w["Region"]="West"
```

```python
season_mapping = {
    "DEC": "Winter",
    "JAN": "Winter",
    "FEB": "Winter",
    "MAR": "Spring",
    "APR": "Spring",
    "MAY": "Spring",
    "JUN": "Summer",
    "JUL": "Summer",
    "AUG": "Summer",
    "SEP": "Autumn",
    "OCT": "Autumn",
    "NOV": "Autumn"
}
precipitation_long_w["SEASON"] = (
    precipitation_long_w["MONTH"].map(season_mapping)
)
```

```python
precipitation_long_e = precipitation_raw.melt(
    id_vars=["YEAR", "LAT", "LON"],
    value_vars=[
        "JAN", "FEB", "MAR", "APR",
        "MAY", "JUN", "JUL", "AUG",
        "SEP", "OCT", "NOV", "DEC"
    ],
    var_name="MONTH",
    value_name="PRECIPITATION"
)
```

```python
precipitation_long_e
```

```python
precipitation_long_e["Region"]="East"
```

```python
season_mapping = {
    "DEC": "Winter",
    "JAN": "Winter",
    "FEB": "Winter",
    "MAR": "Spring",
    "APR": "Spring",
    "MAY": "Spring",
    "JUN": "Summer",
    "JUL": "Summer",
    "AUG": "Summer",
    "SEP": "Autumn",
    "OCT": "Autumn",
    "NOV": "Autumn"
}
precipitation_long_e["SEASON"] = (
    precipitation_long_e["MONTH"].map(season_mapping)
)
```

```python
precipitation_raw_cleaned = pd.concat(
    [precipitation_long_w, precipitation_long_e],
    ignore_index=True
)
```
